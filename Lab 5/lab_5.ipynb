Task 5 – Linear Regression using Gradient Descent

```python
import numpy as np
```

```python
def gradient_descent(X, y, learning_rate=0.01, iterations=1000):

    # Number of training samples (m) and features (n)
    m, n = X.shape

    # Initialize weights with zeros
    weights = np.zeros(n)

    # Initialize bias
    bias = 0

    # Store the cost after each iteration
    cost_history = []

    # Perform Gradient Descent
    for i in range(iterations):

        # Step 1: Predict the output
        y_pred = np.dot(X, weights) + bias

        # Step 2: Calculate prediction error
        error = y_pred - y

        # Step 3: Compute gradients
        dw = (1 / m) * np.dot(X.T, error)
        db = (1 / m) * np.sum(error)

        # Step 4: Update weights and bias
        weights = weights - learning_rate * dw
        bias = bias - learning_rate * db

        # Step 5: Compute Mean Squared Error (Cost)
        cost = (1 / (2 * m)) * np.sum(error ** 2)

        # Store the cost
        cost_history.append(cost)

    return weights, bias, cost_history
```

```python
# Train the Linear Regression model
weights, bias, cost_history = gradient_descent(
    X_train,
    y_train,
    learning_rate=0.01,
    iterations=1000
)
```

Task 6 – Effect of Learning Rate on Gradient Descent

```python
# Define different learning rates
learning_rates = [0.001, 0.01, 0.1]
```

```python
# Dictionary to store cost history for each learning rate
results = {}
```

```python
# Train the model for each learning rate
for lr in learning_rates:

    weights, bias, cost_history = gradient_descent(
        X_train,
        y_train,
        learning_rate=lr,
        iterations=1000
    )

    # Store cost history
    results[lr] = cost_history
```

```python
# Create a new figure
plt.figure(figsize=(8,5))

# Plot the cost curve for each learning rate
for lr in learning_rates:
    plt.plot(results[lr], label=f"Learning Rate = {lr}")

# Add labels and title
plt.xlabel("Iterations")
plt.ylabel("Cost (MSE)")
plt.title("Learning Rate vs Cost Convergence")

# Display legend and grid
plt.legend()
plt.grid(True)

# Show the graph
plt.show()
```

Conclusion

The convergence graph shows that different learning rates produce different optimization behaviors. A learning rate of 0.001 converges slowly due to small parameter updates, while 0.1 reaches the minimum cost much faster. The learning rate of 0.01 provides a balanced trade-off between convergence speed and stability. Selecting an appropriate learning rate is essential for efficient and accurate model training.

Task 7 – Model Evaluation

```python
# Train the model using the best learning rate

weights, bias, cost_history = gradient_descent(
    X_train,
    y_train,
    learning_rate=0.1,
    iterations=1000
)
```

```python
# Predict the final grades for the testing dataset

y_pred = np.dot(X_test, weights) + bias
```

```python
# Import regression evaluation metrics

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
```

```python
# Calculate Mean Absolute Error

mae = mean_absolute_error(y_test, y_pred)
```

```python
# Calculate Mean Squared Error

mse = mean_squared_error(y_test, y_pred)
```

```python
# Calculate Root Mean Squared Error

rmse = np.sqrt(mse)
```

```python
# Calculate R² Score

r2 = r2_score(y_test, y_pred)
```

```python
# Display the evaluation metrics

print("Model Evaluation")
print("-----------------------")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```
