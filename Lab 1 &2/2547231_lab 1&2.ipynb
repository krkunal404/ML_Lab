### Task 2: Missing Value Treatment Strategy Justification

**Strategy for `city` dataset:**

1.  **`Xylene` column:** This column has a very high percentage of missing values (61.32%). Imputing such a large proportion of data could introduce significant bias and distort the underlying distribution. Therefore, we **dropped** this column to maintain data integrity.
2.  **Other numerical columns (`PM2.5`, `PM10`, `NO`, `NO2`, `NOx`, `NH3`, `CO`, `SO2`, `O3`, `Benzene`, `Toluene`, `AQI`):** These columns contain varying but significant numbers of missing values. We chose to **impute** these missing values with the **median** of their respective columns. The median is robust to outliers and is a suitable choice for skewed distributions, which is common for pollution data. This approach retains all rows of data while providing reasonable estimates for the missing sensor readings.
3.  **`AQI_Bucket` column:** This is a categorical column. We chose to **impute** missing values with the **mode** (most frequent value) of the column. This is a standard practice for handling missing categorical data, as it assigns the most probable category.

**Strategy for `crop` dataset:**

1.  **`Production` column:** This is a numerical column with 3730 missing values (1.5% of the total). We chose to **impute** these missing values with the **median** of the column. This approach is preferred over dropping rows, as it retains valuable information from the other columns, and the median is robust to any potential outliers in production figures.

```python
city.drop('Xylene', axis=1, inplace=True)
```

```python
numeric_cols = city.select_dtypes(include=np.number).columns

for col in numeric_cols:
    city[col].fillna(city[col].median(), inplace=True)
```

```python
city['AQI_Bucket'].fillna(
    city['AQI_Bucket'].mode()[0],
    inplace=True
)
```

```python
crop['Production'].fillna(
    crop['Production'].median(),
    inplace=True
)
```

```python
plt.figure(figsize=(10,3))

sns.boxplot(x=city['AQI'])

plt.title("AQI Before Outlier Treatment")

plt.show()
```

```python
Q1 = city['AQI'].quantile(0.25)
Q3 = city['AQI'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
outliers = city[
    (city['AQI'] < lower) |
    (city['AQI'] > upper)
]

print("Number of Outliers:", len(outliers))
```

```python
city['AQI_clean'] = city['AQI'].clip(
    lower=lower,
    upper=upper
)
```

```python
fig, ax = plt.subplots(1,2, figsize=(12,4))

sns.boxplot(x=city['AQI'], ax=ax[0])
ax[0].set_title("Before")

sns.boxplot(x=city['AQI_clean'], ax=ax[1])
ax[1].set_title("After")

plt.show()
```

Outlier Detection Method:
IQR method was used because AQI data is skewed.

Number of Outliers:
3192

Treatment:
Winsorization (capping) was applied.

Reason:
Extreme AQI values may be genuine observations. Capping reduces their influence while preserving all records.

Result:
The boxplot after treatment shows reduced impact of extreme values.

```python
state_aqi_avg = (
    city.groupby('State')['AQI']
    .mean()
    .reset_index()
)

state_aqi_avg.head()
```

```python
crop_state = (
    crop.groupby(['State_Name', 'Crop_Year'])
    .agg({
        'Area': 'sum',
        'Production': 'sum'
    })
    .reset_index()
)

crop_state.head(2)
```

```python
state_aqi_avg = (
    city.groupby('State')['AQI']
    .mean()
    .reset_index()
)

state_aqi_avg.head()
```

```python
state_aqi_avg['State'] = (
    state_aqi_avg['State']
    .str.strip()
    .str.title()
)

crop_state['State_Name'] = (
    crop_state['State_Name']
    .str.strip()
    .str.title()
)
```

```python
merged = pd.merge(
    state_aqi_avg,
    crop_state,
    left_on='State',
    right_on='State_Name',
    how='inner'
)

print("Merged Shape:", merged.shape)

merged.head()
```

```python
plt.figure(figsize=(8,6))

corr_matrix = merged.select_dtypes(include='number').corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Matrix of AQI, Area and Production')

plt.show()
```

**To the Honorable Minister,**

Our analysis of air quality and agricultural data in India reveals several key insights for your consideration.

**Findings:**

1.  **Air Quality Improvement:** India's overall air quality appears to be improving. The average Air Quality Index (AQI) has generally decreased from 2015 to 2020, with 2020 showing the lowest average AQI. This suggests that current environmental policies may be having a positive effect.
2.  **Seasonal Pollution Peaks:** Air pollution is consistently worst during the October-December period, aligning with claims about crop residue burning. This harvest season consistently records significantly higher AQI values compared to other months.
3.  **Weak Link to Crop Production:** We found a very weak negative relationship between average state-level air quality and crop production. States with slightly poorer air quality might have a marginally lower crop output, but this connection is not strong. Factors like rainfall, irrigation, and soil quality likely play a much larger role.

**Recommendation:**

Focus efforts on mitigating air pollution during the October-December harvest season through targeted initiatives to manage agricultural waste, potentially by promoting alternative uses for crop residue or supporting farmers in adopting less polluting practices.

**Limitation:**

While our data shows a correlation, it does not definitively prove that poor air quality *causes* a reduction in crop yields. Many other agricultural and environmental factors influence crop production, and further dedicated studies would be needed to establish a direct cause-and-effect relationship.
